# What is Markdown?
Markdown is a way of writing text that, when read by an iterpreter, formats the text that you've written so that it includes styling such as heading, bold and italics.



# Markdown Text Formating
*When placing asterisks * capsulating a sentence of text the text comes out in italics (just like this one)*

**Double asterisks ** bold the text they incapsulate in markdown**

***Triple asterisks *** bold and italize the text they encapsulate at the same time***
