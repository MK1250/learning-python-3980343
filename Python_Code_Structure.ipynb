# Writing Functions in Python


##Key Terms

**Parameters** : Also called an input parameter, the placeholder inside the function that takes the value of the *argument* being passed to it.

**Argument** : A value or other input that is passed into a function.

**Calling a function** : Telling a function to carry out its task using real values.

**Function** : A code block that performs/defines a task when called.

#### Defining a function:

```
`def function_Name(Parameter_1, Parameter_2):`
```
***def*** is one of *Python's reserverd words*: python reads it as "Function starts here" then defines the name and any parameter the function takes as inputs between parenthesis

Functions do not need parameteres sometimes. They do need () always but not parameters

```
def function_name():
```


````python
from datetime import datetime

# Defining a function (With Parameters):
def Calculate_paycheck(hourly_rate, num_hours):
  paycheck = hourly_rate * num_hours
  return paycheck

# Calling a function
_result = Calculate_paycheck(16.99,24)
print(_result)

#Defining a funtcion (No Parameters):
def TimeOfDay():
  now = datetime.now()
  return now

# Calling a function (Method 2)
print(TimeOfDay())
````

# Writing Pseudocode:

Key terms:

**Pseudocoding** : The process of writing the steps and logic that you would implement in code, buyt in normal language rather than in commands that a programing language could execute.

***Good pseudocode meets the following criteria***
* Any nontechnical person can read your pseudocode and undrstand what you're trying to accomplish.
* Pseudocode can be easily translatable into any programming language, by any programmer.
* Pseudocode can be used as helpful comments in the real code.

***General Guidelines for writing pseudocode***
* Write one instruction or statement per line. (Numbering your lines can help you stay organized.)
* Start with the major steps, then add detail as needed beneath those steps.
* Indent to show hierarchy and blocks within steps.
* Capitalize action words, such as READ, WRITE, REPEAT, IF, THEN, ADD, and SUBTRACT.
* Have someone review your pseudocode to validate the logical flow, ask questions, and provide feedback. This can be a technical or nontechnical person.
